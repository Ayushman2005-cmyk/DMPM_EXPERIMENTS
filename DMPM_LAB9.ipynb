```python
import pandas as pd
from scipy.stats import chi2_contingency

# Observed frequencies from the table
observed = pd.DataFrame(
    [
        [23, 18, 8, 8],
        [15, 6, 12, 10]
    ],
    index=['Male', 'Female'],
    columns=['Strawberry', 'Coffee', 'Orange', 'Vanilla']
)

print("Null Hypothesis (H₀): Preferred flavour of chocolate cream is independent of gender.")
print("Alternative Hypothesis (H₁): Preferred flavour of chocolate cream is dependent on gender (not independent).")

print("Observed Frequency Table:")
display(observed)

# Compute chi-square test parameters
chi2, p, dof, expected = chi2_contingency(observed)

expected_df = pd.DataFrame(expected, index=observed.index, columns=observed.columns)
print("\nCalculated Degrees of Freedom:", dof)
print("\nExpected Frequency Table:")
display(expected_df)
```

```python
# Define the transactional data provided by the user
transactions_data = [
    ['Bread', 'Milk'],
    ['Bread', 'Diaper', 'Beer', 'Eggs'],
    ['Milk', 'Diaper', 'Beer', 'Cake'],
    ['Bread', 'Milk', 'Diaper', 'Beer'],
    ['Bread', 'Milk', 'Diaper', 'Cake']
]

print("Transactions:")
for t in transactions_data:
    print(t)

import pandas as pd
from mlxtend.preprocessing import TransactionEncoder

te = TransactionEncoder()
te_ary = te.fit(transactions_data).transform(transactions_data)
df = pd.DataFrame(te_ary, columns=te.columns_)

print("One-hot encoded DataFrame:")
display(df)

from mlxtend.frequent_patterns import apriori

frequent_itemsets = apriori(df, min_support=0.6, use_colnames=True)

print("Frequent Itemsets (min_support=0.6):")
display(frequent_itemsets)
```
