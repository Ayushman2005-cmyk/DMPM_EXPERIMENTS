```python
#Q1
import numpy as np
data = np.array([10,20,30,40,50])
mean = np.mean(data)
std = np.std(data)
z_score = (data - mean) / std
print(f"Z-score Normalization: {z_score.round(2)}")
min_value = np.min(data)
max_value = np.max(data)
min_max = (data - min_value) / (max_value - min_value)
print(f"Min-Max Normalization: {min_max.round(2)}")
j = len(str(int(max_value)))
decimal_scaling = data / (10 ** j)
print(f"Decimal Scaling Normalization: {decimal_scaling.round(2)}")
```

```python
#Q2
import pandas as pd
import numpy as np
file_path = '/content/Spotify Streaming Performance Dataset.csv'
column_to_normalize = 'Solo Streams (in millions)'
df_simple = pd.read_csv(file_path)
data_to_normalize_simple = df_simple[column_to_normalize].values
max_abs_value_simple = np.max(np.abs(data_to_normalize_simple))
j_simple = len(str(int(max_abs_value_simple)))
decimal_scaling_simple = data_to_normalize_simple / (10 ** j_simple)
df_simple[f'{column_to_normalize}_decimal_scaled_simple'] = decimal_scaling_simple
print("Original data and simplified decimal scaled results:")
display(df_simple[[column_to_normalize, f'{column_to_normalize}_decimal_scaled_simple']].head())
```

```python
#Q3
import pandas as pd
import numpy as np
file_path_mm = '/content/Spotify Streaming Performance Dataset.csv'
column_to_normalize_mm = 'Solo Streams (in millions)'
df_minmax = pd.read_csv(file_path_mm)
data_to_normalize_mm = df_minmax[column_to_normalize_mm].values
min_value_mm = np.min(data_to_normalize_mm)
max_value_mm = np.max(data_to_normalize_mm)
min_max_normalized = (data_to_normalize_mm - min_value_mm) / (max_value_mm - min_value_mm)
df_minmax[f'{column_to_normalize_mm}_min_max_normalized'] = min_max_normalized
print("Original data and Min-Max normalized results (first 5 rows):")
display(df_minmax[[column_to_normalize_mm, f'{column_to_normalize_mm}_min_max_normalized']].head())
```

```python
#Q4
import pandas as pd
import numpy as np
file_path_zs = '/content/Spotify Streaming Performance Dataset.csv'
column_to_normalize_zs = 'Solo Streams (in millions)'
df_zscore = pd.read_csv(file_path_zs)
data_to_normalize_zs = df_zscore[column_to_normalize_zs].values
mean_zs = np.mean(data_to_normalize_zs)
variance_zs = np.var(data_to_normalize_zs)
std_zs = np.std(data_to_normalize_zs)
print(f"Statistics for '{column_to_normalize_zs}':")
print(f"Mean: {mean_zs:.2f}")
print(f"Variance: {variance_zs:.2f}")
print(f"Standard Deviation: {std_zs:.2f}")
zscore_normalized = (data_to_normalize_zs - mean_zs) / std_zs
df_zscore[f'{column_to_normalize_zs}_zscore_normalized'] = zscore_normalized
print("\nOriginal data and Z-score normalized results (first 5 rows):")
display(df_zscore[[column_to_normalize_zs, f'{column_to_normalize_zs}_zscore_normalized']].head())
```
