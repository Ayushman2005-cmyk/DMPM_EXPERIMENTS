```python
a = int(input("Enter the number of observations:"))
total = 0
for i in range(a):
    num = float(input(f"Enter observation {i+1}: "))
    total += num
mean = total / a
print(f"The mean of the data points is: {mean}")
```

```python
def calculate_average(numbers):
    if not numbers:
        return 0
    total = sum(numbers)
    count = len(numbers)
    return total / count
count = int(input("How many datapoints do you have? "))
data_points = []
for i in range(count):
    val = float(input(f"Enter datapoint {i+1}: "))
    data_points.append(val)
average_value = calculate_average(data_points)
print(f"The average of {data_points} is: {average_value}")
```

```python
nums = input("Enter numbers separated by spaces: ")
numbers = [float(x) for x in nums.split()]
numbers.sort()
n = len(numbers)
if n % 2 == 1:
    median = numbers[n // 2]
else:
    median = (numbers[(n // 2) - 1] + numbers[n // 2]) / 2
print("Sorted numbers:", numbers)
print("Median:", median)
```

```python
def find_median(numbers):
    numbers.sort()
    n = len(numbers)
    mid = n // 2
    if n % 2 == 1:
        return numbers[mid]
    else:
        return (numbers[mid - 1] + numbers[mid]) / 2
nums = input("Enter numbers separated by spaces: ")
data = [float(x) for x in nums.split()]
print(f"Sorted data: {sorted(data)}")
print(f"The median is: {find_median(data)}")
```

```python
nums = input("Enter numbers separated by spaces: ")
numbers = [float(x) for x in nums.split()]
count = 0
mode = numbers[0]
for x in numbers:
    count1 = numbers.count(x)
    if count1 > count:
        count = count1
        mode = x
print(f"The mode is: {mode} (appeared {count} times)")
```

```python
def find_mode(numbers):
    if not numbers:
        return None
    counts = {}
    for num in numbers:
        counts[num] = counts.get(num, 0) + 1
    return max(counts, key=counts.get)
nums = input("Enter numbers separated by spaces (e.g., 1 2 2 3 4): ")
data_points = [float(x) for x in nums.split()]
if data_points:
    mode_value = find_mode(data_points)
    print(f"Data points entered: {data_points}")
    if mode_value is not None:
        print(f"The mode is: {mode_value}")
    else:
        print("No mode could be determined (e.g., empty input).")
else:
    print("No data points were provided.")
```

```python
nums = input("Enter numbers separated by spaces (e.g., 10 20 30): ")
data_points = [float(x) for x in nums.split()]
if len(data_points) < 1:
    print("Cannot calculate variance with no data points.")
else:
    total_sum = 0
    for x in data_points:
        total_sum += x
    mean = total_sum / len(data_points)
    sum_sq_diff = 0
    for x in data_points:
        sum_sq_diff += (x - mean) ** 2
    variance = sum_sq_diff / len(data_points)
    print(f"Data points: {data_points}")
    print(f"Mean: {mean:.2f}")
    print(f"Variance: {variance:.2f}")
```

```python
def calculate_variance(numbers):
    if not numbers:
        return 0.0
    mean = sum(numbers) / len(numbers)
    sum_sq_diff = sum((x - mean) ** 2 for x in numbers)
    variance = sum_sq_diff / len(numbers)
    return variance
num_input = input("Enter numbers separated by spaces (e.g., 10 20 30): ")
data_points = [float(x) for x in num_input.split()]
variance_value = calculate_variance(data_points)
print(f"Data points: {data_points}")
print(f"The variance is: {variance_value:.2f}")
```

```python
import math
nums = input("Enter numbers separated by spaces (e.g., 1 2 3): ")
data_points = [float(x) for x in nums.split()]
if len(data_points) < 1:
    print("No data points provided.")
else:
    total_sum = sum(data_points)
    mean = total_sum / len(data_points)
    sum_sq_diff = 0
    for x in data_points:
        sum_sq_diff += (x - mean) ** 2
    variance = sum_sq_diff / len(data_points)
    std_dev = math.sqrt(variance)
    print(f"Data points: {data_points}")
    print(f"Mean: {mean:.2f}")
    print(f"Variance: {variance:.2f}")
    print(f"Standard Deviation (without function): {std_dev:.2f}")
```

```python
import math
def calculate_std_dev(numbers):
    if not numbers:
        return 0.0
    mean = sum(numbers) / len(numbers)
    sum_sq_diff = sum((x - mean) ** 2 for x in numbers)
    variance = sum_sq_diff / len(numbers)
    return math.sqrt(variance)
nums = input("Enter numbers separated by spaces (e.g., 1 2 3): ")
data_points = [float(x) for x in nums.split()]
if not data_points:
    print("No data points provided.")
else:
    std_dev_value = calculate_std_dev(data_points)
    print(f"Data points: {data_points}")
    print(f"Standard Deviation (with function): {std_dev_value:.2f}")
```
