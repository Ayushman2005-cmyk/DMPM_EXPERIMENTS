```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
arr1 = np.array([1,2,3,4,5])
arr2 = np.array([6, 7, 8, 9, 10])
```

```python
plt.figure(figsize=(8, 6))
sns.boxplot(x=arr1)
plt.title('Box Plot of arr1')
plt.xlabel('Values')
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.histplot(arr1, bins=5, kde=True)
plt.title('Histogram of arr1')
plt.xlabel('Values')
plt.ylabel('Frequency')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(x=arr1, y=arr2)
plt.title('Scatter Plot of arr1 vs arr2')
plt.xlabel('arr1 Values')
plt.ylabel('arr2 Values')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

```python
plt.figure(figsize=(8, 6))
plt.plot(np.arange(len(arr1)), arr1, marker='o', linestyle='-')
plt.title('Line Chart of arr1')
plt.xlabel('Index')
plt.ylabel('Values')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

```python
categories = ['A', 'B', 'C', 'D', 'E']
values = [10, 25, 15, 30, 20]
plt.figure(figsize=(8, 6))
sns.barplot(x=categories, y=values, palette='viridis')
plt.title('Bar Chart of Categories')
plt.xlabel('Category')
plt.ylabel('Value')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
labels = ['Category X', 'Category Y', 'Category Z']
sizes = [40, 30, 30]
explode = (0.1, 0, 0)
plt.figure(figsize=(8, 8))
plt.pie(sizes, explode=explode, labels=labels, autopct='%1.1f%%', shadow=True, startangle=90)
plt.title('Pie Chart of Proportions')
plt.axis('equal')
plt.show()
```
