```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import pandas as pd
df = pd.read_csv('/content/BMW sales data (2010-2024).csv')
display(df.head())
print(df.columns)
```

```python
print(df.columns)
```

```python
plt.figure(figsize=(8, 6))
sns.boxplot(x=df['Price_USD'])
plt.title('Box Plot of Price (USD)')
plt.xlabel('Price (USD)')
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.histplot(df['Price_USD'], bins=20, kde=True)
plt.title('Histogram of Price (USD)')
plt.xlabel('Price (USD)')
plt.ylabel('Frequency')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(x=df['Price_USD'], y=df['Sales_Volume'])
plt.title('Scatter Plot of Price (USD) vs Sales Volume')
plt.xlabel('Price (USD)')
plt.ylabel('Sales Volume')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

```python
plt.figure(figsize=(12, 6))
sales_by_year = df.groupby('Year')['Sales_Volume'].sum().reset_index()
plt.plot(sales_by_year['Year'], sales_by_year['Sales_Volume'], marker='o', linestyle='-')
plt.title('Total Sales Volume Over Years')
plt.xlabel('Year')
plt.ylabel('Sales Volume')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

```python
classification_counts = df['Sales_Classification'].value_counts().sort_index()
plt.figure(figsize=(10, 6))
sns.barplot(x=classification_counts.index, y=classification_counts.values, hue=classification_counts.index, palette='viridis', legend=False)
plt.title('Distribution of Sales Classification')
plt.xlabel('Sales Classification')
plt.ylabel('Number of Sales')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
region_counts = df['Region'].value_counts()
labels = region_counts.index
sizes = region_counts.values
explode = [0] * len(sizes)
if len(sizes) > 0:
    largest_slice_index = sizes.argmax()
    explode[largest_slice_index] = 0.1
plt.figure(figsize=(8, 8))
plt.pie(sizes, explode=explode, labels=labels, autopct='%1.1f%%', shadow=True, startangle=90)
plt.title('Proportion of Sales by Region')
plt.axis('equal')
plt.show()
```
