```python
#Q1
import numpy as np

data = np.array([10, 20, 30, 40, 50])

z = (data - np.mean(data)) / np.std(data)
print("Z-score:", z.round(2))

min_max = (data - np.min(data)) / (np.max(data) - np.min(data))
print("Min-Max:", min_max.round(2))

j = len(str(np.max(data)))
decimal = data / (10 ** j)
print("Decimal Scaling:", decimal.round(2))
```

```python
#Q2
import pandas as pd
import numpy as np

df = pd.read_csv('/content/Spotify Streaming Performance Dataset.csv')

column = 'Solo Streams (in millions)'
data = df[column]

j = len(str(int(np.max(np.abs(data)))))
df['Decimal Scaled'] = data / (10 ** j)

print(df[[column, 'Decimal Scaled']].head())
```

```python
#Q3
import pandas as pd
import numpy as np

df = pd.read_csv('/content/Spotify Streaming Performance Dataset.csv')

column = 'Solo Streams (in millions)'
data = df[column]

df['Min-Max'] = (data - np.min(data)) / (np.max(data) - np.min(data))

print(df[[column, 'Min-Max']].head())
```

```python
#Q4
import pandas as pd
import numpy as np

df = pd.read_csv('/content/Spotify Streaming Performance Dataset.csv')

column = 'Solo Streams (in millions)'
data = df[column]

mean = np.mean(data)
std = np.std(data)

df['Z-score'] = (data - mean) / std

print("Mean:", round(mean, 2))
print("Standard Deviation:", round(std, 2))
print(df[[column, 'Z-score']].head())
```
